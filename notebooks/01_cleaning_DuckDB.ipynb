```python
con.execute("""
    UPDATE retail_clean AS r
    SET ProductNameClean = nb_occurrences.ProductNameClean
    FROM (
        SELECT
            StockCode,
            ProductNameClean,
            RANK() OVER (PARTITION BY StockCode ORDER BY COUNT(*) DESC) AS rang_frequence
        FROM retail_clean
        WHERE ProductNameClean IS NOT NULL
        GROUP BY StockCode, ProductNameClean
    ) AS nb_occurrences
    WHERE r.StockCode = nb_occurrences.StockCode
      AND nb_occurrences.rang_frequence = 1
      AND r.StockCode NOT IN (
          SELECT StockCode FROM (
              SELECT StockCode, RANK() OVER (PARTITION BY StockCode ORDER BY COUNT(*) DESC) AS rang_frequence
              FROM retail_clean WHERE ProductNameClean IS NOT NULL
              GROUP BY StockCode, ProductNameClean
          ) sub WHERE rang_frequence = 1
          GROUP BY StockCode HAVING COUNT(*) > 1
      )
""")
```

```python
con.execute("""
    CREATE OR REPLACE TABLE DIM_DATE AS
    SELECT
        ROW_NUMBER() OVER (ORDER BY InvoiceDate) AS DateID,
        InvoiceDate AS Date,
        EXTRACT(YEAR FROM InvoiceDate) AS Year,
        EXTRACT(QUARTER FROM InvoiceDate) AS Quarter,
        EXTRACT(MONTH FROM InvoiceDate) AS Month,
        STRFTIME(InvoiceDate, '%B') AS MonthName,
        EXTRACT(WEEK FROM InvoiceDate) AS WeekOfYear,
        EXTRACT(DAY FROM InvoiceDate) AS Day,
        STRFTIME(InvoiceDate, '%A') AS DayName,
        EXTRACT(DOW FROM InvoiceDate) AS DayOfWeek,
        CASE WHEN EXTRACT(DOW FROM InvoiceDate) IN (0, 6) THEN TRUE ELSE FALSE END AS IsWeekend
    FROM (
        SELECT DISTINCT InvoiceDate
        FROM retail_clean
        WHERE InvoiceDate IS NOT NULL
    )
    ORDER BY InvoiceDate
""")

con.sql("SELECT * FROM DIM_DATE LIMIT 5").fetchdf()
```

```python
con.execute("""
    CREATE OR REPLACE TABLE FACT_SALES AS
    SELECT
        r.OrderLineID,
        r.Invoice,
        d.DateID,
        p.ProductID,
        c.CustomerKey,
        g.GeographyID,
        w.WarehouseKey,
        oc.OrderContextID,
        r.Quantity,
        r.UnitPriceGBP,
        r.UnitCostGBP,
        r.DeliveryTime,
        r.Quantity * TRY_CAST(r.UnitPriceGBP AS DOUBLE) AS SalesAmount
    FROM retail_clean r
    LEFT JOIN DIM_DATE d
        ON r.InvoiceDate = d.Date
    LEFT JOIN DIM_PRODUCT p
        ON r.StockCode = p.StockCode
    LEFT JOIN DIM_CUSTOMER c
        ON r.CustomerID = c.CustomerID
    LEFT JOIN DIM_GEOGRAPHY g
        ON r.Country = g.Country
    LEFT JOIN DIM_WAREHOUSE w
        ON r.WarehouseID = w.WarehouseID
    LEFT JOIN DIM_ORDER_CONTEXT oc
        ON r.SalesChannel IS NOT DISTINCT FROM oc.SalesChannel
        AND r.PaymentMethod IS NOT DISTINCT FROM oc.PaymentMethod
        AND r.PromotionCode IS NOT DISTINCT FROM oc.PromotionCode
        AND r.DiscountRate IS NOT DISTINCT FROM oc.DiscountRate
        AND r.OrderStatus IS NOT DISTINCT FROM oc.OrderStatus
""")

con.sql("SELECT * FROM FACT_SALES LIMIT 10").fetchdf()
```
